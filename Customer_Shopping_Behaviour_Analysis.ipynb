```python
import sqlite3

conn = sqlite3.connect("customer_shopping.db")

cleaned_df.to_sql(
    "customer_shopping",
    conn,
    if_exists="replace",
    index=False
)

print("Database and table created successfully!")
```

```python
query = """
SELECT 
    "Item Purchased",
    SUM("Purchase Amount (USD)") AS Total_Sales
FROM customer_shopping
GROUP BY "Item Purchased"
ORDER BY Total_Sales DESC
LIMIT 10;
"""

top_products = pd.read_sql_query(query, conn)

top_products
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=top_products,
    x='Total_Sales',
    y='Item Purchased'
)

plt.title('Top 10 Products by Total Sales')
plt.xlabel('Total Sales (USD)')
plt.ylabel('Item Purchased')
plt.show()
```

```python
query = """
SELECT 
    Category,
    COUNT(*) AS Total_Purchases,
    SUM("Purchase Amount (USD)") AS Total_Sales,
    ROUND(AVG("Purchase Amount (USD)"), 2) AS Average_Purchase
FROM customer_shopping
GROUP BY Category
ORDER BY Total_Sales DESC;
"""

category_sql = pd.read_sql_query(query, conn)

category_sql
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=category_sql,
    x='Category',
    y='Total_Sales'
)

plt.title('Total Sales by Category - SQL Analysis')
plt.xlabel('Category')
plt.ylabel('Total Sales (USD)')
plt.show()
```

```python
query = """
SELECT
    "Subscription Status",
    COUNT(*) AS Total_Customers,
    SUM("Purchase Amount (USD)") AS Total_Sales,
    ROUND(AVG("Purchase Amount (USD)"), 2) AS Average_Purchase
FROM customer_shopping
GROUP BY "Subscription Status"
ORDER BY Total_Sales DESC;
"""

subscription_sql = pd.read_sql_query(query, conn)

subscription_sql
```

```python
plt.figure(figsize=(7,5))

sns.barplot(
    data=subscription_sql,
    x='Subscription Status',
    y='Average_Purchase'
)

plt.title('Average Purchase by Subscription Status - SQL')
plt.xlabel('Subscription Status')
plt.ylabel('Average Purchase (USD)')
plt.show()
```

```python
query = """
SELECT
    "Discount Applied",
    COUNT(*) AS Total_Purchases,
    SUM("Purchase Amount (USD)") AS Total_Sales,
    ROUND(AVG("Purchase Amount (USD)"), 2) AS Average_Purchase
FROM customer_shopping
GROUP BY "Discount Applied"
ORDER BY Total_Sales DESC;
"""

discount_sql = pd.read_sql_query(query, conn)

discount_sql
```

```python
plt.figure(figsize=(7,5))

sns.barplot(
    data=discount_sql,
    x='Discount Applied',
    y='Average_Purchase'
)

plt.title('Average Purchase by Discount Status')
plt.xlabel('Discount Applied')
plt.ylabel('Average Purchase (USD)')
plt.show()
```

```python
query = """
SELECT
    "Payment Method",
    COUNT(*) AS Total_Purchases,
    SUM("Purchase Amount (USD)") AS Total_Sales,
    ROUND(AVG("Purchase Amount (USD)"), 2) AS Average_Purchase
FROM customer_shopping
GROUP BY "Payment Method"
ORDER BY Total_Purchases DESC;
"""

payment_sql = pd.read_sql_query(query, conn)

payment_sql
```

```python
plt.figure(figsize=(9,5))

sns.barplot(
    data=payment_sql,
    x='Payment Method',
    y='Total_Purchases'
)

plt.title('Purchases by Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Number of Purchases')
plt.xticks(rotation=45)
plt.show()
```

```python
query = """
SELECT
    Location,
    COUNT(*) AS Total_Purchases,
    SUM("Purchase Amount (USD)") AS Total_Sales,
    ROUND(AVG("Purchase Amount (USD)"), 2) AS Average_Purchase
FROM customer_shopping
GROUP BY Location
ORDER BY Total_Sales DESC
LIMIT 10;
"""

location_sql = pd.read_sql_query(query, conn)

location_sql
```

```python
plt.figure(figsize=(10,6))

sns.barplot(
    data=location_sql,
    x='Total_Sales',
    y='Location'
)

plt.title('Top 10 Locations by Total Sales')
plt.xlabel('Total Sales (USD)')
plt.ylabel('Location')
plt.show()
```

```python
query = """
SELECT
    CASE
        WHEN "Previous Purchases" <= 10 THEN 'Low'
        WHEN "Previous Purchases" <= 30 THEN 'Medium'
        ELSE 'High'
    END AS Customer_Type,
    COUNT(*) AS Total_Customers,
    ROUND(AVG("Purchase Amount (USD)"), 2) AS Average_Purchase
FROM customer_shopping
GROUP BY Customer_Type
ORDER BY Average_Purchase DESC;
"""

repeat_sql = pd.read_sql_query(query, conn)

repeat_sql
```

```python
plt.figure(figsize=(7,5))

sns.barplot(
    data=repeat_sql,
    x='Customer_Type',
    y='Average_Purchase'
)

plt.title('Average Purchase by Customer Type')
plt.xlabel('Customer Type')
plt.ylabel('Average Purchase (USD)')
plt.show()
```
